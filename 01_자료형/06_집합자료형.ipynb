```python
# 집합자료형은 어떻게 만들까?
## 집합자료형의 특징
## 1. 중복 허용하지 않음
## 2. 순서가 없다

s1 = set([1, 2, 3])
s2 = set("hello")
s3 = {1, 2, 3}
s4 = {}
s5 = set()

print(s1, type(s1))
print(s2, type(s2))
print(s3, type(s3))
print(s4, type(s4))
print(s5, type(s5))
```

```python
## 교집합
s1 = {1, 2, 3, 4, 5, 6}
s2 = {4, 5, 6, 7, 8, 9}

print(s1 & s2)
print(s1.intersection(s2)) ## 교집합 intersection
```

```python
## 합집합
s1 = {1, 2, 3, 4, 5, 6}
s2 = {4, 5, 6, 7, 8, 9}

print(s1 | s2)
```
