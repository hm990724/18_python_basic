```python
# 리스트는 어떻게 만들고 사용할까?

# 리스트명 = [요소1, 요소2 ...]

a = []
b = [1, 2, 3]
c = ["Life", "is", "too", "short"]
d = [1, 2, ["life", "is"]]
e = [1, 2, "life", "is"]
f = list()

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
a[2]
a[-1]
```

```python
## 리스트의 슬라이싱

a = [1, 2, 3, 4, 5]

a[1:4]
```

```python
a = [1, 2, 3, ["a", "b", "c"]]

a[3][1:]
```

```python
## 리스트의 연산

a = [1, 2, 3]
b = [4, 5, 6]

a + b
```

```python
a * 3
```

```python
## 리스트 길이 구하기 - len

a = [1, 2, 3]

len(a)
```

```python
## 리스트의 수정과 삭제

a = [1, 2, 3]

a[2] = 4
```

```python
a
```

```python
a = [1, 2, 3]

del a[1]

print(a)
```

```python
## 리스트 관련 함수

## 리스트에 요소 추가 하기 - append

a = [1, 2, 3]

a.append(4)
print(a)
```

```python
a.append(["a", "b"])

print(a)
```
