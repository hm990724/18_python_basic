```python
## 딕셔너리 쌍 추가하기

a = {1: "a"}
a[2] = "b"
#a = {"name": "홍길동"}


print(a)
```

```python
a[3] = [1, 2, 3]
print(a)
```

```python
a = {
    1: "a",
    2: "b",
    "name": "홍길동",
    3 : [1, 2, 3],
    (1, 2): (1, 2),
    #[1, 2]: [1, 2]

}

a=3
print(a)
```

```python
## 딕셔너리에서 key를 사용해서 value 값 얻기

학점 = {
    "홍길동": 10,
    "유재석": 90
}
print(학점["유재석"])
print(학점["홍길동"])

학점["홍길동"] = 80
학점["강호동"] = 70

print(학점)
```

```python
## 딕셔너리 관련 함수
## key리스트 만들기 - keys()

dict = {
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02"
}

dict.keys()
list(dict.keys())

```

```python
## value 리스트 만들기 - values()
dict = {
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02" 

}
dict.values()
list(dict.values())
```

```python
## key, value 쌍 얻기 - items

dict = {
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02" 

}
dict.items()
list(dict.items())
```

```python
## key, Value 상 모두 지우기 - clear

dict = {
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02" 

}

print(dict)
dict.clear()
print(dict)
```

```python
## key로 value값 얻기 - get

dict = {
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02" 

}
dict.get("name")
# dict["name"]

dict.get("hobby")
#dict["hobby"] error
```
