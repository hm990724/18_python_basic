```python
## 문자열 길이 구하기 len()

a = "Life is too short"
len(a) # len("Life is too short")
```

```python
## 문자열 인덱싱과 슬라이싱
a = "Life is too short"
a[3]
a[16]
a[8]
a[-1]
a[-2]
```

```python
## 슬라이싱
a = "Life is too short"
a[0:3+1]
a[5:7]
a[-5:-1]
a[-5: ]
a[:4]
```

```python
## 문자열 포메팅

a = 3
b = "five"

print("I eat %d apples" % a)
print("I eat %d apples" % b)
```

```python
## format() 함수를 사용한 포메팅

"I eat {0} apples".format(3) 
"I eat {0} apples".format("five") 
```

```python
## 2개 이상의 값을 넣을때

number = 10
day = "three"

"I ate {0} apples, so i was sick for {1} days".format(number, day)
```

```python
## f문자열 포메팅

name = "홍길동"
age = 30

print(type(name))
print(type(age))

print(f"나의 이름은 {name}이고, 나이는 {age}입니다.")
```

```python
## 문자열 관련 함수들

a = "hobby"
type(a)

## 문자 개수 세기 - count

a.count("b")
# a.count("b") == "hobby".count(a)
```

```python
## 위치 알려주기 .find()

a = "Python is the best choice"
print(a.find("b"))
print(a.find("k"))
```

```python
## 위치 알려주기2 -- index()

a = "Life is too short"
print(a.index("t"))
a.index("k")
```

```python
## 문자열 삽입 - join

", ".join(["a", "b", "c", "d"])
```

```python
## 소문자를 대문자로 바꾸기 - upper

a = "hi"
a.upper()
```
