```python
# 문자열은 어떻게 만들고 사용할까? 

## 1. 큰 따옴표
print("Hello World")

## 2. 작은 따옴표
print('Python is fun')

## 3. 큰따옴표 3개를 연속으로 감싸기 
print("""Life is too short, You need Python""")

## 4. 작은따옴표 3개를 연속으로 감싸기 
print('''Life is too short, You need Python''')
```

```python
## 문자열 길이 구하기 len()

a = "Life is too short"
len(a) # len("Life is too short")
```

```python
## 문자열 인덱싱과 슬라이싱
a = "Life is too short"
a[3]
a[16]
a[8]
a[-1]
a[-2]
```

```python
## 슬라이싱
a = "Life is too short"
a[0:3+1]
a[5:7]
a[-5:-1]
a[-5: ]
a[:4]
```
